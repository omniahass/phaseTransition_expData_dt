```python
# ROI measurement- plot multiple maps on the same figure

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from dipy.io.image import load_nifti, save_nifti

patient_directory = '/Users/gabriellebaxter/Documents/HEAL_Volunteers/WCMyofascial'
vol = '2005'
parent_directory = patient_directory + vol + '/'
sides = ['left','right']

diffusion_times = ['0022ms','0042ms','0081ms','0156ms','0300ms']

fig, axes = plt.subplots(1, 2, figsize=(15, 5))

blues = [
    "#084594",  # dark blue
    "#2171b5",
    "#4292c6",
    "#6baed6",
    "#9ecae1"   # light blue
]

# 5 orange shades
oranges = [
    "#7f2704",  # dark orange/brown
    "#a63603",
    "#d94801",
    "#f16913",
    "#fd8d3c"   # light orange
]

map_configs = ['_greg_header','_greg_dcm2nii','_jellescript_header','_jellescript_dcm2nii','_jellescript_bmatrix']
for i in range(0,len(sides)):
    side = sides[i]

    for j in range(len(map_configs)):
        maps_config = map_configs[j]
        map_directory = os.path.join(parent_directory,'maps'+maps_config) # normal
        roi_file = os.path.join(parent_directory,side + '.nii.gz') #+ side + '.nii.gz'
        csv_output_filename = parent_directory + 'dti_params_diffusion_time_vol' + vol + '_' + side + maps_config + '.csv'
        
        available_maps = os.listdir(map_directory)
        print(available_maps)

        roi, roi_affine, roi_img = load_nifti(roi_file, return_img=True)
        roi = np.asarray(roi)
        print(np.shape(roi))
        nx,ny,nz = np.shape(roi)
        
        dti_params = ['RD','AD']
        dti_param_values = ['RD_mean','RD_std','RD_min','RD_max','AD_mean','AD_std','AD_min','AD_max']

        column_names = ['diffusion_time'] + dti_param_values
        df = pd.DataFrame(columns = column_names)

        for diffusion_time in diffusion_times:
            relevant_maps = [s for s in available_maps if diffusion_time in s]
            average_params = []

            for param in dti_params:
                matching_map = [s for s in relevant_maps if param in s]
                matching_map.sort()
                current_map_file = os.path.join(map_directory,matching_map[-1])
                print(current_map_file)
                current_map, map_affine, map_img = load_nifti(current_map_file, return_img=True)
                masked_map = current_map
                masked_map[roi == 0] = 0
                roi_only = masked_map[masked_map != 0]
                roi_only = roi_only[~np.isnan(roi_only)]
                average_params.append(np.median(roi_only)) # mean
                average_params.append(np.std(roi_only)) # std
                average_params.append(np.min(roi_only)) # min
                average_params.append(np.max(roi_only))
            temp_data = [diffusion_time] + average_params
            temp_df = pd.DataFrame([temp_data],columns = column_names)
            df = pd.concat([df,temp_df])
            df.to_csv(csv_output_filename,index=False)

            diff_times = [float(s[:-2]) for s in diffusion_times]

        # Plot AD and RD
        axes[i].errorbar(diff_times,df['AD_mean'], yerr=df['AD_std'],color = blues[j])
        axes[i].plot(diff_times,df['AD_mean'],'.',linestyle='--',marker='o',color = blues[j])
        axes[i].errorbar(diff_times,df['RD_mean'], yerr=df['RD_std'],color = oranges[j])
        axes[i].plot(diff_times,df['RD_mean'],'.',linestyle='--',marker='o',color=oranges[j])
        axes[i].set_xlabel('t (ms)')
        axes[i].set_ylabel(r'$\mu m^2 / ms$')
        axes[i].set_title(side)
        axes[i].set_ylim(0.3,2.5)
        # axes[i].legend(['AD','RD'])

labels = [
    "AD_greg_header", "RD_greg_header",
    "AD_greg_dcm2nii", "RD_greg_dcm2nii",
    "AD_jellescript_header", "RD_jellescript_header",
    "AD_jellescript_dcm2nii", "RD_jellescript_dcm2nii",
    "AD_jellescript_bmatrix", "RD_jellescript_bmatrix"
]

fig.legend(labels, loc='center left', bbox_to_anchor=(1, 0.5))
figure_filename = parent_directory + 'vol' + vol + '_vectorcomparison.png'
plt.savefig(figure_filename,dpi=200,bbox_inches='tight')
plt.show()
```
